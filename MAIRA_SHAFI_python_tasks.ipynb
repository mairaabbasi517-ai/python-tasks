```python
# task 1 cinema ticket price
name = input("enter your name: ")
age = int(input("enter your age:"))
if age < 5 or age > 100:
    price = 0
elif age >= 5 and age <= 12:
    price = 300
elif age >=13 and age <= 59:
    price = 700
elif age >= 60 and age <= 100:
    price = 400
print("costomer name:", name)
print("final ticket price:", price)
```

```python
#temperature converter
celsius = float(input("enter temperature in celsius: "))

fahrenhiet = (celsius * 9/5) + 32

print("temperature in fahrenhiet:", fahrenhiet)

if celsius < 10:
  print("weather condition:very cold")
elif celsius < 20:
  print("weather condition:cold")
elif celsius < 30:
  print("weather condition:normal")
else:
  print("weather condition:hot")
```

```python
#mobile recharege calculator

mobile_number = input("enter mobile number: ")
recharge_amount = float(input("enter recharge amount : "))

tax = recharge_amount * 3 / 100
final_balance = recharge_amount - tax

print("mobile number:", mobile_number)
print("tax:", tax)
print("final received balance:", final_balance)
```

```python
secret_pin = "1234"

for attempt in range(3):
    pin = input("Enter PIN: ")

    if pin == secret_pin:
        print("Access Granted")
        break
    else:
        remaining = 2 - attempt
        print("Incorrect PIN")
        print("Remaining attempts:", remaining)

else:
    print("Account Blocked")
```

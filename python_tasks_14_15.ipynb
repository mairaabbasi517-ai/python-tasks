```python
#TASK 11
expenses = []

total_expenses = int(input("Enter total expenses: "))

for i in range(total_expenses):
    print("\nExpense", i + 1)

    name = input("Enter expense name: ")
    category = input("Enter category (Food/Transport/Shopping/Bills/Other): ")
    amount = float(input("Enter amount: "))

    expense = {
        "name": name,
        "category": category,
        "amount": amount
    }

    expenses.append(expense)

budget = float(input("\nEnter your budget: "))


def final_report():
    total = 0
    category_spending = {
        "Food": 0,
        "Transport": 0,
        "Shopping": 0,
        "Bills": 0,
        "Other": 0
    }

    highest_expense = expenses[0]

    for expense in expenses:
        total += expense["amount"]
        category_spending[expense["category"]] += expense["amount"]

        if expense["amount"] > highest_expense["amount"]:
            highest_expense = expense

    print("\n--- Final Report ---")
    print("Total Spending:", total)

    print("\nCategory Spending:")
    for category, amount in category_spending.items():
        print(category, ":", amount)

    print("\nHighest Expense:")
    print(highest_expense["name"], ":", highest_expense["amount"])

    if total > budget:
        print("\nWarning: Budget Exceeded!")
    else:
        print("\nBudget is within limit.")


final_report()
```

```python
#TASK 12
inventory = {}


def add_product():
    name = input("Enter product name: ")
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))

    if name in inventory:
        print("Product already exists!")
    else:
        inventory[name] = {
            "price": price,
            "quantity": quantity
        }
        print("Product added successfully!")


def sell_product():
    name = input("Enter product name: ")

    if name in inventory:
        quantity = int(input("Enter quantity to sell: "))

        if quantity <= inventory[name]["quantity"]:
            inventory[name]["quantity"] -= quantity
            print("Product sold successfully!")
        else:
            print("Insufficient stock!")
    else:
        print("Product not found!")


def restock_product():
    name = input("Enter product name: ")

    if name in inventory:
        quantity = int(input("Enter quantity to add: "))
        inventory[name]["quantity"] += quantity
        print("Product restocked successfully!")
    else:
        print("Product not found!")


def search_product():
    name = input("Enter product name: ")

    if name in inventory:
        print("Price:", inventory[name]["price"])
        print("Quantity:", inventory[name]["quantity"])
    else:
        print("Product not found!")


def display_inventory():
    print("\n--- Inventory ---")

    total_value = 0

    for name, product in inventory.items():
        print(name, "- Price:", product["price"],
              "Quantity:", product["quantity"])

        if product["quantity"] < 5:
            print("Low Stock!")

        total_value += product["price"] * product["quantity"]

    print("Total Inventory Value:", total_value)


while True:
    print("\n1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid choice!")
```

```python
#TASK 13
books = []


def add_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)
    print("Book added successfully!")


def display_books():
    if not books:
        print("No books available.")
        return

    for book in books:
        status = "Available" if book["available"] else "Borrowed"

        print("\nBook ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", status)


def search_book():
    title = input("Enter book title: ")

    for book in books:
        if book["title"].lower() == title.lower():
            print("Book Found!")
            print("Book ID:", book["id"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Borrowed")
            return

    print("Book not found!")


def borrow_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:
            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed!")
            return

    print("Book not found!")


def return_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:
            if not book["available"]:
                book["available"] = True
                print("Book returned successfully!")
            else:
                print("Book is already available!")
            return

    print("Book not found!")


while True:
    print("\n1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        borrow_book()
    elif choice == "5":
        return_book()
    elif choice == "6":
        print("Goodbye!")
        break
    else:
        print("Invalid choice!")
```

```python
#TASK 14: RESTURANRT ORDERING SYSTEM
menu = {"burger":500,
       "pizza":1200,
       "pasta":800,
       "drink":200}

orders = []

print("========= resturant menu ==========")
for key, value in menu.items():
   print(f'{key} : {value}')

while True:
    item_name = input(" Enter your item name: ").title()
    if item_name in menu:
      quantity = int(input("Enter quantity of your item: "))
      price = menu[item_name]
      sub_total = price * quantity

      orders.append({
            "items": item_name,
            "quantity": quantity,
            "sub_total": sub_total
     })
      print("item added successfully!")
      print("sub_total:",sub_total)

      more = input("do you want to order another item? (yes/no):").lower()
      if more != "yes":
        break
      else:
          print("sorry!this item is not available in menu.")

#calculate total bill
total_bill = 0

for order in orders:total_bill =+ order["sub_total"]

#calculate 13% tax
tax = total_bill * 0.13

#calculate 10 discount if bill is Rs 3000 or above
if total_bill >= 3000:
      discount = total_bill * 0.10
else:
      discount = 0

#calculate final amount
grand_total = total_bill + tax - discount

#receipt function
def print_receipt(orders, total_bill, tax, discount, grand_total):

    print("\n")
    print("==========FINAL RECEIPT==========")
    for order in orders:
        print("item:",order["items"])
        print("quantity:",order["quantity"])
        print("subtotal:",order["sub_total"])
        print("----------------------------")
        print("total bill:",total_bill)
        print("tax (13%):",tax)
        print("discount:",discount)
        print("grand total:",grand_total)
        print("---------------------------")
        print_receipt(orders, total_bill, tax, discount, grand_total)
```

```python
#task 15

patients = []

def add_patient():
    id = input("Patient ID: ")

    for p in patients:
        if p["id"] == id:
            print("ID already exists!")
            return

    patient = {
        "id": id,
        "name": input("Name: "),
        "age": input("Age: "),
        "type": input("Type (normal/emergency): ").lower()
    }

    patients.append(patient)
    print("Patient added!")


def search_patient():
    id = input("Enter Patient ID: ")

    for p in patients:
        if p["id"] == id:
            print(p)
            return

    print("Patient not found!")


def display_patients():
    print("\nEmergency Patients:")
    for p in patients:
        if p["type"] == "emergency":
            print(p)

    print("\nNormal Patients:")
    for p in patients:
        if p["type"] == "normal":
            print(p)


def cancel_appointment():
    id = input("Enter Patient ID: ")

    for p in patients:
        if p["id"] == id:
            p["appointment"] = "Cancelled"
            print("Appointment cancelled!")
            return

    print("Patient not found!")


def count_patients():
    normal = sum(p["type"] == "normal" for p in patients)
    emergency = sum(p["type"] == "emergency" for p in patients)

    print("Normal:", normal)
    print("Emergency:", emergency)


while True:
    print("\n1. Add Patient")
    print("2. Search Patient")
    print("3. Display Patients")
    print("4. Cancel Appointment")
    print("5. Patient Count")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_patient()
    elif choice == "2":
        search_patient()
    elif choice == "3":
        display_patients()
    elif choice == "4":
        cancel_appointment()
    elif choice == "5":
        count_patients()
    elif choice == "6":
        break
    else:
        print("Invalid choice!")
```
